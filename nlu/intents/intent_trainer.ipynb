## Train clair_intents

```python
# Initialize the classifier
dataset_name = 'clair_intents'
clf = IntentClassifier(examples_file=f'data/{dataset_name}_examples.yml', 
                       dataset_name=dataset_name,
                       handle_punctuation=True)

# Setup training parameters
clf.config.sent_hl_units = 16
clf.config.sent_dropout = 0.1
clf.config.learning_rate = 0.0005
clf.config.epochs = 1000
clf.config.callback_patience = 100
clf.config.validation_split = 0.05
clf.config.wandb_project = 'intent-classifier'
# Start training
clf.train(save_model=f'models/{dataset_name}')
```

### Load cases from a file and test and inspect outputs

```python

# cases = []
data_file_path = "M:/Datasets/Chats-NL-Enzymes/preprocessed/chats.csv"   
n_groups = 5

import pandas as pd
import numpy as np
def read_data(data_file_path: str):
    try:
        data_df = pd.read_csv(data_file_path, index_col=0)
    except:
        data_df = pd.read_csv(data_file_path, index_col=0, sep='|')
    data_df = data_df.drop(columns=['L1', 'L2C'], errors='ignore')
    data_df = data_df.rename(columns={
        'message':'text', 'dialog_id': 'group'
    }, errors='ignore')
    data_df['text'] = data_df['text'].fillna(' ')
    return data_df


data_df = read_data(data_file_path)
if 0 < n_groups < len(data_df['group'].unique()):
    groups = np.random.choice(data_df['group'].unique(), n_groups, replace=False)
    data_df = data_df[data_df['group'].isin(groups)]
data_df = data_df.sort_values(by='timestamp')
data_df.head()
```

```python
cases = []
for item in data_df['text']:
    certainties = clf.predict_one(item,get_certainty="all")[1] 
    cases.append(certainties)

cases = pd.DataFrame(cases)

data_df = data_df.join(cases)

# Export to excel
writer = pd.ExcelWriter(data_file_path.replace('.csv', '__TESTING_SSIM.xlsx'), engine='xlsxwriter')
for i, group in enumerate(data_df['group'].unique()):
    data_df[data_df['group'] == group].to_excel(writer, encoding="utf-8", sheet_name=f"Chat-{i+1}")
writer.save()

data_df.head()
```
